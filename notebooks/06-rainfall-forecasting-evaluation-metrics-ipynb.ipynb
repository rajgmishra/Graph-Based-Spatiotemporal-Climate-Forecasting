```python
import os
import numpy as np
import pandas as pd

DATA_DIR = (
    "/kaggle/input/datasets/drrajgauravmishra/spatiotemporal-gnn-kaggle-data/Spatiotemporal_GNN_Kaggle_Data"
)

test_data = np.load(
    os.path.join(DATA_DIR, "testing_annual.npz")
)

Y_test = test_data["Y"]
Y_test_mask = test_data["Y_mask"]
target_years = test_data["target_years"]

print("TEST DATASET INSPECTION")
print("=" * 50)

print("Target shape:", Y_test.shape)
print("Mask shape:", Y_test_mask.shape)
print("Target years:", target_years)

rainfall = Y_test[:, :, 0]
rainfall_mask = Y_test_mask[:, :, 0]

print("\nRAINFALL STATISTICS")
print("=" * 50)

for i, year in enumerate(target_years):
    valid = rainfall_mask[i]
    values = rainfall[i, valid]

    print(f"\nYear: {year}")
    print("Valid observations:", len(values))
    print("Missing observations:", np.sum(~valid))
    print("Minimum:", np.min(values))
    print("Maximum:", np.max(values))
    print("Mean:", np.mean(values))
    print("Median:", np.median(values))
```

```python
thresholds = [1.0, 2.5, 5.0]

results = []

for i, year in enumerate(target_years):

    valid_values = rainfall[i, rainfall_mask[i]]
    total_valid = len(valid_values)

    for threshold in thresholds:

        event_count = np.sum(valid_values >= threshold)
        non_event_count = total_valid - event_count

        results.append({
            "Year": int(year),
            "Threshold_mm_day": threshold,
            "Event_nodes": int(event_count),
            "Non_event_nodes": int(non_event_count),
            "Event_percentage": 100 * event_count / total_valid
        })

# Pooled test-set frequencies
pooled_values = rainfall[rainfall_mask]

for threshold in thresholds:

    event_count = np.sum(pooled_values >= threshold)
    total_valid = len(pooled_values)

    results.append({
        "Year": "Pooled",
        "Threshold_mm_day": threshold,
        "Event_nodes": int(event_count),
        "Non_event_nodes": int(total_valid - event_count),
        "Event_percentage": 100 * event_count / total_valid
    })

threshold_frequency = pd.DataFrame(results)

print(threshold_frequency.round(2).to_string(index=False))
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Load test data
test_data = np.load(os.path.join(DATA_DIR, "testing_annual.npz"))

X_test = torch.tensor(test_data["X"], dtype=torch.float32, device=device)
Y_test = test_data["Y"]
Y_test_mask = test_data["Y_mask"]
target_years = test_data["target_years"]

target_mean = np.array([3.0588877, 31.36195], dtype=np.float32)
target_std = np.array([2.156659, 2.990062], dtype=np.float32)

print("Device:", device)
print("Test input shape:", X_test.shape)
print("Target years:", target_years)


# GRU architecture
class GRUBaseline(nn.Module):
    def __init__(self, input_dim=4, hidden_dim=128, output_dim=2, dropout=0.2):
        super().__init__()
        self.gru = nn.GRU(
            input_size=input_dim,
            hidden_size=hidden_dim,
            num_layers=1,
            batch_first=True
        )
        self.dropout = nn.Dropout(dropout)
        self.fc = nn.Linear(hidden_dim, output_dim)

    def forward(self, x):
        batch_size, time_steps, num_nodes, num_features = x.shape
        x = x.permute(0, 2, 1, 3)
        x = x.reshape(batch_size * num_nodes, time_steps, num_features)

        _, hidden = self.gru(x)
        representation = self.dropout(hidden[-1])
        output = self.fc(representation)

        return output.reshape(batch_size, num_nodes, 2)


# Generate predictions for each seed
gru_predictions = {}

for seed in [42, 123, 2023]:

    checkpoint_path = os.path.join(
        GRU_DIR, f"gru_best_seed_{seed}.pt"
    )

    model = GRUBaseline().to(device)
    state_dict = torch.load(
        checkpoint_path, map_location=device, weights_only=False
    )
    model.load_state_dict(state_dict)
    model.eval()

    with torch.no_grad():
        pred_norm = model(X_test).cpu().numpy()

    pred_raw = pred_norm * target_std + target_mean
    gru_predictions[seed] = pred_raw

    print(
        f"Seed {seed}: Prediction shape = {pred_raw.shape}, "
        f"Finite = {np.isfinite(pred_raw).all()}"
    )
```

```python
# LSTM architecture
class LSTMBaseline(nn.Module):
    def __init__(self, input_dim=4, hidden_dim=128, output_dim=2, dropout=0.2):
        super().__init__()
        self.lstm = nn.LSTM(
            input_size=input_dim,
            hidden_size=hidden_dim,
            num_layers=1,
            batch_first=True
        )
        self.dropout = nn.Dropout(dropout)
        self.fc = nn.Linear(hidden_dim, output_dim)

    def forward(self, x):
        batch_size, time_steps, num_nodes, num_features = x.shape
        x = x.permute(0, 2, 1, 3)
        x = x.reshape(batch_size * num_nodes, time_steps, num_features)

        _, (hidden, _) = self.lstm(x)
        representation = self.dropout(hidden[-1])
        output = self.fc(representation)

        return output.reshape(batch_size, num_nodes, 2)


lstm_predictions = {}

for seed in [42, 123, 2023]:

    checkpoint_path = os.path.join(
        LSTM_CHECKPOINT_DIR,
        f"lstm_h128_e300_seed{seed}.pt"
    )

    checkpoint = torch.load(
        checkpoint_path, map_location=device, weights_only=False
    )

    model = LSTMBaseline().to(device)
    model.load_state_dict(checkpoint["model_state_dict"])
    model.eval()

    with torch.no_grad():
        pred_norm = model(X_test).cpu().numpy()

    pred_raw = pred_norm * target_std + target_mean
    lstm_predictions[seed] = pred_raw

    print(
        f"Seed {seed}: Prediction shape = {pred_raw.shape}, "
        f"Finite = {np.isfinite(pred_raw).all()}"
    )
```

```python
# GATv2-GRU architecture
class GATv2GRUForecaster(nn.Module):
    def __init__(self, input_dim=256, hidden_dim=64, output_dim=2, dropout=0.2):
        super().__init__()
        self.gru = nn.GRU(
            input_size=input_dim,
            hidden_size=hidden_dim,
            num_layers=1,
            batch_first=True
        )
        self.dropout = nn.Dropout(dropout)
        self.fc = nn.Linear(hidden_dim, output_dim)

    def forward(self, x):
        batch_size, time_steps, num_nodes, num_features = x.shape
        x = x.permute(0, 2, 1, 3)
        x = x.reshape(batch_size * num_nodes, time_steps, num_features)

        _, hidden = self.gru(x)
        representation = self.dropout(hidden[-1])
        output = self.fc(representation)

        return output.reshape(batch_size, num_nodes, 2)


gatv2_gru_predictions = {}

for seed in [42, 123, 2023]:

    # Load the matching spatial embeddings
    embedding_path = os.path.join(
        EMBEDDING_INPUT_DIR,
        f"GATv2_seed_{seed}_embeddings_2000_2024.npy"
    )

    all_embeddings = np.load(embedding_path)

    # Test target years 2024 and 2025 use input years 2014-2023 and 2015-2024
    X_test_embedding = np.stack([
        all_embeddings[14:24],
        all_embeddings[15:25]
    ])

    X_test_embedding = torch.tensor(
        X_test_embedding, dtype=torch.float32, device=device
    )

    # Load matching forecasting checkpoint
    checkpoint_path = os.path.join(
        GATV2_GRU_CHECKPOINT_DIR,
        f"GATv2_GRU_H64_E500_Seed{seed}.pt"
    )

    checkpoint = torch.load(
        checkpoint_path, map_location=device, weights_only=False
    )

    model = GATv2GRUForecaster().to(device)
    model.load_state_dict(checkpoint["model_state_dict"])
    model.eval()

    with torch.no_grad():
        pred_norm = model(X_test_embedding).cpu().numpy()

    pred_raw = pred_norm * target_std + target_mean
    gatv2_gru_predictions[seed] = pred_raw

    print(
        f"Seed {seed}: Prediction shape = {pred_raw.shape}, "
        f"Finite = {np.isfinite(pred_raw).all()}"
    )
```

```python
thresholds = [2.5, 5.0]

def calculate_csi_hss(observed, predicted, mask, threshold):

    obs_event = observed[mask] >= threshold
    pred_event = predicted[mask] >= threshold

    hits = np.sum(obs_event & pred_event)
    misses = np.sum(obs_event & ~pred_event)
    false_alarms = np.sum(~obs_event & pred_event)
    correct_negatives = np.sum(~obs_event & ~pred_event)

    csi_denominator = hits + misses + false_alarms
    csi = hits / csi_denominator if csi_denominator > 0 else np.nan

    hss_denominator = (
        (hits + misses) * (misses + correct_negatives)
        + (hits + false_alarms) * (false_alarms + correct_negatives)
    )

    hss = (
        2 * (hits * correct_negatives - misses * false_alarms)
        / hss_denominator
        if hss_denominator > 0 else np.nan
    )

    return {
        "CSI": csi,
        "HSS": hss,
        "Hits": hits,
        "Misses": misses,
        "False Alarms": false_alarms,
        "Correct Negatives": correct_negatives
    }


model_predictions = {
    "GRU": gru_predictions,
    "LSTM": lstm_predictions,
    "GATv2-GRU": gatv2_gru_predictions
}

results = []

for model_name, seed_predictions in model_predictions.items():

    for seed, predictions in seed_predictions.items():

        for threshold in thresholds:

            # Individual test years
            for year_index, year in enumerate(target_years):

                scores = calculate_csi_hss(
                    Y_test[year_index, :, 0],
                    predictions[year_index, :, 0],
                    Y_test_mask[year_index, :, 0],
                    threshold
                )

                results.append({
                    "Model": model_name,
                    "Seed": seed,
                    "Year": int(year),
                    "Threshold": threshold,
                    **scores
                })

            # Pooled scores: combine contingency counts first
            pooled_scores = calculate_csi_hss(
                Y_test[:, :, 0].reshape(-1),
                predictions[:, :, 0].reshape(-1),
                Y_test_mask[:, :, 0].reshape(-1),
                threshold
            )

            results.append({
                "Model": model_name,
                "Seed": seed,
                "Year": "Pooled",
                "Threshold": threshold,
                **pooled_scores
            })


csi_hss_results = pd.DataFrame(results)

print("Total result rows:", len(csi_hss_results))
display(csi_hss_results.round(4))
```

```python
csi_hss_summary = (
    csi_hss_results
    .groupby(["Model", "Year", "Threshold"])
    .agg(
        CSI_mean=("CSI", "mean"),
        CSI_std=("CSI", "std"),
        HSS_mean=("HSS", "mean"),
        HSS_std=("HSS", "std")
    )
    .reset_index()
)

for col in ["CSI_mean", "CSI_std", "HSS_mean", "HSS_std"]:
    csi_hss_summary[col] = csi_hss_summary[col].round(4)

display(csi_hss_summary)

# Save results
csi_hss_results.to_csv(
    "/kaggle/working/csi_hss_results_by_seed.csv",
    index=False
)

csi_hss_summary.to_csv(
    "/kaggle/working/csi_hss_summary.csv",
    index=False
)

print("\nBoth CSV files saved successfully.")
```

```python
import numpy as np
import torch
from sklearn.neighbors import BallTree

# Reverse coordinate columns to reproduce the original parser
coords_reversed = coordinates[:, [1, 0]]
coords_reversed_rad = np.radians(coords_reversed)

# Reconstruct graph exactly as in the original notebook
tree_reversed = BallTree(
    coords_reversed_rad,
    metric="haversine"
)

_, indices_reversed = tree_reversed.query(
    coords_reversed_rad,
    k=9
)

edge_index_reversed = []

for i, neighbours in enumerate(indices_reversed):
    for j in neighbours[1:]:
        edge_index_reversed.append([i, j])

edge_index_reversed = torch.tensor(
    edge_index_reversed,
    dtype=torch.long
).t().contiguous()

# Compare with the saved graph
original_edges_set = set(map(
    tuple,
    graph_data.edge_index.cpu().numpy().T.tolist()
))

reversed_edges_set = set(map(
    tuple,
    edge_index_reversed.numpy().T.tolist()
))

common = original_edges_set.intersection(reversed_edges_set)

print("Original edges:", len(original_edges_set))
print("Reconstructed edges:", len(reversed_edges_set))
print("Matching edges:", len(common))
print("Exact edge match:", original_edges_set == reversed_edges_set)
```

```python
import numpy as np
import pandas as pd
from sklearn.metrics.pairwise import haversine_distances

# Original and reconstructed edge sets
original_edge_array = graph_data.edge_index.cpu().numpy().T
reconstructed_edge_array = edge_index_reversed.cpu().numpy().T

original_edges_set = set(map(tuple, original_edge_array.tolist()))
reconstructed_edges_set = set(map(tuple, reconstructed_edge_array.tolist()))

# Identify mismatched edges
missing_edges = original_edges_set - reconstructed_edges_set
additional_edges = reconstructed_edges_set - original_edges_set

print("Original-only edges:", len(missing_edges))
print("Reconstructed-only edges:", len(additional_edges))

# Calculate geographical distances using reversed coordinates
earth_radius_km = 6371.0

def edge_distance_km(src, dst):
    return (
        haversine_distances(
            coords_reversed_rad[src].reshape(1, -1),
            coords_reversed_rad[dst].reshape(1, -1)
        )[0, 0] * earth_radius_km
    )

# Display mismatches
records = []

for (src, dst) in sorted(missing_edges):
    records.append({
        "source": src,
        "original_neighbour": dst,
        "original_distance_km": edge_distance_km(src, dst),
        "type": "Original only"
    })

for (src, dst) in sorted(additional_edges):
    records.append({
        "source": src,
        "original_neighbour": dst,
        "original_distance_km": edge_distance_km(src, dst),
        "type": "Reconstructed only"
    })

mismatch_df = pd.DataFrame(records)

print("\nMismatch details:")
display(mismatch_df.sort_values(["source", "type"]).reset_index(drop=True))

print("\nDistance statistics:")
display(
    mismatch_df.groupby("type")["original_distance_km"]
    .agg(["count", "min", "median", "mean", "max"])
)
```

```python
import numpy as np
import pandas as pd
from sklearn.metrics.pairwise import haversine_distances

# Correct coordinates: latitude, longitude
coords_correct_rad = np.radians(coordinates)

# Original graph edges
original_edges = graph_data.edge_index.cpu().numpy()

def geographic_edge_distances(edge_index, coords_rad):
    src, dst = edge_index
    distances = haversine_distances(
        coords_rad[src],
        coords_rad[dst]
    ).diagonal() * 6371.0
    return distances

# Distances of original graph edges using correct coordinates
original_distances = geographic_edge_distances(
    original_edges, coords_correct_rad
)

# Correct geographic KNN graph
from sklearn.neighbors import BallTree

tree_correct = BallTree(coords_correct_rad, metric="haversine")
_, correct_indices = tree_correct.query(
    coords_correct_rad, k=9
)

correct_edges = np.array([
    (i, j)
    for i, neighbours in enumerate(correct_indices)
    for j in neighbours[1:]
]).T

correct_distances = geographic_edge_distances(
    correct_edges, coords_correct_rad
)

comparison = pd.DataFrame({
    "Graph": ["Original graph", "Correct geographic KNN"],
    "Mean distance (km)": [
        original_distances.mean(),
        correct_distances.mean()
    ],
    "Median distance (km)": [
        np.median(original_distances),
        np.median(correct_distances)
    ],
    "Maximum distance (km)": [
        original_distances.max(),
        correct_distances.max()
    ]
})

display(comparison)
```
